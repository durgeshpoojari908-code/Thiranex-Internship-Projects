```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 300

# Generate employee details
departments = np.random.choice(
    ["IT", "HR", "Finance", "Sales"], n
)

genders = np.random.choice(
    ["Male", "Female"], n
)

ages = np.random.randint(22, 46, n)

join_dates = pd.to_datetime(
    np.random.choice(
        pd.date_range("2018-01-01", "2025-12-31"),
        n
    )
)

# Department-wise base salary
department_salary = {
    "IT": 55000,
    "HR": 40000,
    "Finance": 60000,
    "Sales": 45000
}

base_salary = np.array(
    [department_salary[d] for d in departments]
)

# Calculate experience
experience = np.clip(
    2026 - join_dates.year.to_numpy(),
    0,
    8
)

# Generate salary
salaries = (
    base_salary
    + (ages - 22) * 1000
    + experience * 1500
    + np.random.normal(0, 8000, n)
).round(2)

# Create DataFrame
ml_df = pd.DataFrame({
    "Name": [f"Employee_{i+1:03d}" for i in range(n)],
    "Age": ages,
    "Salary": salaries,
    "Department": departments,
    "Gender": genders,
    "Join_Date": join_dates
})

# Create target variable
median_salary = ml_df["Salary"].median()

ml_df["Salary_Level"] = np.where(
    ml_df["Salary"] >= median_salary,
    "High",
    "Low"
)

# Save dataset
ml_df.to_csv(
    r"C:\Users\Rupesh.Amin\Thiranex_Task_2\Dataset\employee_ml_synthetic.csv",
    index=False
)

# Display results
print("Synthetic Dataset Created Successfully!")
print("Dataset Shape:", ml_df.shape)

print("\nSalary Level Distribution:")
print(ml_df["Salary_Level"].value_counts())

display(ml_df.head())
```

```python
# Create experience feature
ml_df["Experience"] = np.clip(
    2026 - pd.to_datetime(ml_df["Join_Date"]).dt.year,
    0,
    8
)

# Select input features
X = ml_df[
    ["Age", "Experience", "Department", "Gender"]
]

# Select target
y = ml_df["Salary_Level"]

# Convert categorical columns into numeric values
X = pd.get_dummies(
    X,
    columns=["Department", "Gender"],
    drop_first=True,
    dtype=int
)

print("Preprocessing Completed!")
print("\nFeature Shape:", X.shape)
print("Target Shape:", y.shape)

display(X.head())
```

```python
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Scale the features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Train-Test Split Completed!")

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

print("Training Labels:", y_train.shape)
print("Testing Labels:", y_test.shape)
```

```python
# Initialize models
models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "Decision Tree": DecisionTreeClassifier(
        random_state=42,
        max_depth=5
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        max_depth=5
    )
}

# Train and evaluate models
results = {}

for name, model in models.items():

    if name == "Logistic Regression":
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
    else:
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    results[name] = {
        "Model": model,
        "Predictions": y_pred,
        "Accuracy": accuracy
    }

    print(f"{name}: {accuracy * 100:.2f}%")

print("\nAll Models Trained Successfully!")
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, data) in zip(axes, results.items()):

    cm = confusion_matrix(
        y_test,
        data["Predictions"],
        labels=["Low", "High"]
    )

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Low", "High"]
    )

    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(name)

plt.tight_layout()

plt.savefig(
    r"C:\Users\Rupesh.Amin\Thiranex_Task_2\Visualizations\confusion_matrices.png",
    dpi=300
)

plt.show()
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

plt.figure(figsize=(9, 6))

for name, data in results.items():

    model = data["Model"]

    if name == "Logistic Regression":
        probabilities = model.predict_proba(X_test_scaled)[:, 1]
    else:
        probabilities = model.predict_proba(X_test)[:, 1]

    y_binary = (y_test == "High").astype(int)

    fpr, tpr, thresholds = roc_curve(
        y_binary,
        probabilities
    )

    auc_score = roc_auc_score(
        y_binary,
        probabilities
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC = {auc_score:.2f})"
    )

plt.plot([0, 1], [0, 1], "k--", label="Random Classifier")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.grid()

plt.savefig(
    r"C:\Users\Rupesh.Amin\Thiranex_Task_2\Visualizations\roc_curve.png",
    dpi=300
)

plt.show()
```
